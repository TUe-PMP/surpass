# Relative PL intensity during POx/AlOx deposition and annealing

This SciencePlots-styled notebook creates a publication-oriented figure of the relative photoluminescence intensity of n- and p-type InP. The PL intensity is normalized to the corresponding bare surface before POx/AlOx deposition. This version uses math-rendered degree symbols, tilted state labels, and a y-axis starting at zero.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    import scienceplots
    plt.style.use(['science', 'notebook', 'no-latex'])
except ImportError:
    # Install with: pip install SciencePlots
    # The fallback keeps the notebook executable in a minimal environment.
    plt.style.use('default')

output_dir = Path('output')
output_dir.mkdir(exist_ok=True)
```

## Experimental data

The bare-surface point is the normalization reference and is therefore set to unity. Its displayed uncertainty is set to zero by default. Set `BARE_REL_SIGMA` to a nonzero value if an uncertainty for the normalization measurement should be shown.

```python
BARE_REL_SIGMA = 0.0

annealing_data = pd.DataFrame([
    # state                  T (C)  n ratio   p ratio  independent relative 1-sigma
    ('Before deposition',     np.nan, 1.000000, 1.00000, BARE_REL_SIGMA, BARE_REL_SIGMA),
    ('As-deposited',           25.0, 0.919870, 0.74597, 0.05, 0.05),
    ('150 C anneal',          150.0, 1.286678, 0.69940, 0.05, 0.05),
    ('200 C anneal',          200.0, 1.775588, 0.65660, 0.05, 0.05),
    ('250 C anneal',          250.0, 2.825580, 0.59693, 0.05, 0.05),
    ('300 C anneal',          300.0, 1.848540, 0.57730, 0.05, 0.05),
    ('350 C anneal',          350.0, 1.187900, 0.55610, 0.05, 0.05),
], columns=[
    'state', 'temperature_c', 'ratio_n', 'ratio_p',
    'independent_rel_sigma_n', 'independent_rel_sigma_p',
])

annealing_data
```

## Publication figure

```python
plt.rcParams.update({
    'font.size': 11,
    'axes.labelsize': 12,
    'axes.linewidth': 1.25,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'xtick.major.width': 1.1,
    'ytick.major.width': 1.1,
    'xtick.major.size': 5,
    'ytick.major.size': 5,
    'legend.fontsize': 10,
    'savefig.dpi': 600,
})

x = np.arange(len(annealing_data))
labels = [
    'Bare',
    'As-deposited',
    r'150$\,^{\circ}$C',
    r'200$\,^{\circ}$C',
    r'250$\,^{\circ}$C',
    r'300$\,^{\circ}$C',
    r'350$\,^{\circ}$C',
]

fig, ax = plt.subplots(figsize=(7.2, 4.5))

ax.plot(
    x, annealing_data['ratio_n'],
    color='tab:blue', marker='o', markersize=7.0,
    markerfacecolor='tab:blue', markeredgecolor='white', markeredgewidth=0.7,
    linewidth=2.0,
    label='n-InP', zorder=3,
)

ax.plot(
    x, annealing_data['ratio_p'],
    color='tab:red', marker='s', markersize=6.7,
    markerfacecolor='tab:red', markeredgecolor='white', markeredgewidth=0.7,
    linewidth=2.0,
    label='p-InP', zorder=3,
)

ax.axhline(1.0, color='0.45', linewidth=1.0, linestyle=':', zorder=1)
ax.axvline(0.5, color='0.75', linewidth=0.9, linestyle='--', zorder=1)

ax.set_ylabel(r'Relative PL intensity, $I_{\mathrm{PL}}/I_{\mathrm{PL,bare}}$')
ax.set_xlabel('Processing state')
ax.set_xticks(x, labels, rotation=25, ha='right', rotation_mode='anchor')
ax.set_xlim(-0.35, len(x) - 0.65)
ax.set_ylim(0.0, 3.1)
ax.legend(frameon=False, loc='upper right')
ax.tick_params(axis='both', which='both', direction='in',
               top=True, bottom=True, left=True, right=True)
for spine in ax.spines.values():
    spine.set_visible(True)
    spine.set_linewidth(1.25)

fig.savefig(output_dir / 'inp_relative_pl_annealing_curve.png', bbox_inches='tight')
fig.savefig(output_dir / 'inp_relative_pl_annealing_curve.pdf', bbox_inches='tight')
fig.savefig(output_dir / 'inp_relative_pl_annealing_curve.svg', bbox_inches='tight')
plt.show()
```

The dashed vertical line separates the bare reference surface from the states after POx/AlOx deposition.
